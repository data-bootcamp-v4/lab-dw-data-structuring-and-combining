## Challenge 1: Combining & Cleaning Data

In this challenge, we will be working with the customer data from an insurance company, as we did in the two previous labs. The data can be found here:
- https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv

But this time, we got new data, which can be found in the following 2 CSV files located at the links below.

- https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file2.csv
- https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file3.csv

Note that you'll need to clean and format the new data.

Observation:
- One option is to first combine the three datasets and then apply the cleaning function to the new combined dataset
- Another option would be to read the clean file you saved in the previous lab, and just clean the two new files and concatenate the three clean datasets

```python
#importing data
import pandas as pd
file1 = pd.read_csv('https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv')
file2 = pd.read_csv('https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file2.csv')
file3 = pd.read_csv('https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file3.csv')

display(file1)
print("\n")
display(file2)
print("\n")
display(file3)
```

```python
file1.columns = [col.lower().replace(' ', '_') for col in file1.columns]
file2.columns = [col.lower().replace(' ', '_') for col in file2.columns]
file3.columns = [col.lower().replace(' ', '_') for col in file3.columns]

# Rename 'state' to 'st' in file3 for consistency
file3 = file3.rename(columns={'state': 'st'})

# Drop rows with all NaN values as they seem to be artifacts from the CSVs
file1 = file1.dropna(how='all')
file2 = file2.dropna(how='all')
file3 = file3.dropna(how='all')

# Concatenate the three dataframes
combined_df = pd.concat([file1, file2, file3], ignore_index=True)

display(combined_df.head())
print(f"Combined DataFrame shape: {combined_df.shape}")
```

```python
# Checking for duplicates in the new and combined DataFrame
duplicates = combined_df.duplicated()

# Display the rows that are marked as duplicates
duplicate_rows = combined_df[duplicates]

print("Duplicate Rows:")
print(duplicate_rows)

# --- New code to find the origin of duplicates ---
print("\nChecking the origin of these duplicate rows:")

duplicate_customers = duplicate_rows['customer'].tolist()

for customer_id in duplicate_customers:
    in_file1 = customer_id in file1['customer'].values
    in_file2 = customer_id in file2['customer'].values
    in_file3 = customer_id in file3['customer'].values

    files_found_in = []
    if in_file1: files_found_in.append('file1')
    if in_file2: files_found_in.append('file2')
    if in_file3: files_found_in.append('file3')

    print(f"Customer {customer_id} is found in: {', '.join(files_found_in)}")
```

```python
#Drop the duplicates
combined_df = combined_df.drop_duplicates()
combined_df.shape
```

2. Create a pivot table that shows the average customer lifetime value per gender and education level. Analyze the resulting table to draw insights.

```python
file4_pivot = file4.pivot_table(index=['gender', 'education'], values='customer_lifetime_value', aggfunc='mean').reset_index().round(2)

#Slicing it for better understanding it
male_data = file4_pivot[file4_pivot['gender'] == 'M']
female_data = file4_pivot[file4_pivot['gender'] == 'F']

display(male_data)
print("\n")
display(female_data)

#Gender Trends: The CLV for males and females is relatively comparable overall, with each gender taking the lead in different education levels. This suggests that education has a more nuanced effect on CLV than gender alone.
#Education Impact: Engagement varies more significantly across education levels than between genders. Strategies tailored to education-based personas might be more effective than gender-focused strategies in optimizing lifetime value.
```

## Bonus

You work at the customer service department and you want to know which months had the highest number of complaints by policy type category. Create a summary table showing the number of complaints by policy type and month.
Show it in a long format table.

*In data analysis, a long format table is a way of structuring data in which each observation or measurement is stored in a separate row of the table. The key characteristic of a long format table is that each column represents a single variable, and each row represents a single observation of that variable.*

*More information about long and wide format tables here: https://www.statology.org/long-vs-wide-data/*

```python
from typing import dataclass_transform
complaints_per_policy = file4.groupby(['policy_type', 'month'])['number_of_open_complaints'].count().reset_index()(name='complaints_count')
complaints_per_policy
```

```python
#FOR FUN, to understand the difference between table
complaints_per_policy_wide = complaints_per_policy.pivot(index='policy_type', columns='month', values='complaints_count').reset_index()
complaints_per_policy_wide.columns.name = None # Remove the column name 'month'
display(complaints_per_policy_wide)
```

In this wide format table:

*   Each row still represents a `policy_type`.
*   However, the `month` values (1 and 2) are now spread across different columns (`1` and `2`), with the `complaints_count` for each month populating those columns.

This format can sometimes be easier for quick comparisons of specific values (e.g., how many complaints were there for 'Personal Auto' in month 1 versus month 2), but it can become cumbersome with many unique values in the column you're trying to pivot, as it would lead to many columns.
