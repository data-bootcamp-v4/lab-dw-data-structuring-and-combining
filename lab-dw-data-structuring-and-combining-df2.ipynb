```python
# Your code goes here
import pandas as pd

_pd2 = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file2.csv")
df2 = _pd2.copy()
_pd2.head(5)
```

```python
# Formating the column names
df2 = df2.rename(columns={col:col.replace(' ','_').lower() for col in df2.columns})
df2
```

```python
# rename the column st
df2.rename(columns={'st':'state'}, inplace=True)
```

```python
# display the indexes of the columns to change the order
for i, col in enumerate(df2.columns): print(i, col)
```

```python
df2.columns
```

```python
# order the columns in the same order than df1
def move_column(df2, old_pos, new_pos): 
    col = df2.columns[old_pos] 
    df2.insert(new_pos, col, df2.pop(col)) 
    return df2
# move the column in 8 position to the 10
move_column(df2, 8, 10)   
```

```python
# display the indexes of the columns to change the order
for i, col in enumerate(df2.columns): print(i, col)
```

```python
df2.gender.unique()
```

```python
# standardizing Gender,
# mapping gender column
gender_mapping = {
    'female': 'F',
    'Male': 'M'
}
df2.gender = df2.gender.replace(gender_mapping)
```

```python
df2.gender.unique()
```

```python
df2.gender = df2.gender.fillna(df2.gender.mode()[0])
```

```python
list(df2.state.unique())
```

```python
df2.rename(columns={'AZ': 'Arizona'}, inplace=True)
```

```python
df2.education.unique()
```

```python
 # replace "Bachelors"by "Bachelor" in education
education_map ={'Bachelors':'Bachelor'}
df2.education = df2.education.replace(education_map)
```

```python
# In vehicle class, "Sports Car", "Luxury SUV" and "Luxury Car" replaced by "Luxury"
list(df2.vehicle_class.unique())
```

```python
vehicle_class_map ={
    'Luxury SUV':'Luxury',
    'Sports Car':'Luxury',
    'Luxury Car':'Luxury',
}
df2.vehicle_class = df2.vehicle_class.replace(vehicle_class_map)
```

```python
df2.customer_lifetime_value.unique()
```

```python
# delete the % character in Customer Lifetime Value
df2['customer_lifetime_value'] = df2['customer_lifetime_value'].str.replace('%','', regex=True) 

# convert columns type
#df['customer_lifetime_value'] = df['customer_lifetime_value'].astype(float)
```

```python
# convert Number of open complaints
df2.number_of_open_complaints.unique()
```

```python
# Apply the transformation of Number of Open Complaints values from str to int 

df2["number_of_open_complaints"] = (
    df2["number_of_open_complaints"]
    .astype(str)
    # Use .str.extract(r'regex') to extracts the digit(s) between the two slashes
    .str.extract(r'/(\d+)/', expand=False) 
    # Handles NaN values that might occur if the format is absent
    .fillna(0) 
    # Converts the result to an integer
    .astype(int)
)
```

#### Dealing with Null values

```python
# Display the columns with the number of nan
df2.isna().sum()
```

```python
df2.customer_lifetime_value = df2.customer_lifetime_value.astype(float)
```

```python
df2.customer_lifetime_value = df2.customer_lifetime_value.median() 
```

```python
df2.describe()
```

```python
df2.dtypes
```

```python
# convert all the numeric variables to integers
numerical_values = ['customer_lifetime_value',
                   'income', 
                   'monthly_premium_auto', 
                   'total_claim_amount']

df2[numerical_values]= df2[numerical_values].astype(int)
```

#### Dealing with duplicated

```python
df2.duplicated().sum()
```

```python
# Dropping all duplicate rows
df2.drop_duplicates(inplace=True)
```

```python
df2.shape
```

```python
# reset the index to ensure consistency, after dropping duplicates
df2_without_duplicates = df2.copy()
df2_without_duplicates = df2_without_duplicates.drop_duplicates().reset_index(drop=True)
```
