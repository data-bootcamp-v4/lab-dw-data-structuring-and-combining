```python

import pandas as pd

# Datensätze aus GitHub laden
df1 = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv")

df2 = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file2.csv")

df3 = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file3.csv")

# Anzahl der Zeilen und Spalten anzeigen
print("Datei 1:", df1.shape)
print("Datei 2:", df2.shape)
print("Datei 3:", df3.shape)
```

```python

# Drei Datensätze zusammenführen
df = pd.concat([df1, df2, df3], ignore_index=True)

# Größe des zusammengeführten Datensatzes
print("Gesamter Datensatz:", df.shape)

# Die ersten 5 Zeilen anzeigen
df.head()
```

```python

# Spaltennamen in allen drei Dateien vereinheitlichen
for data in [df1, df2, df3]:
    data.columns = data.columns.str.strip().str.lower()
    data.rename(columns={"st": "state"}, inplace=True)

# Datensätze erneut zusammenführen
df = pd.concat([df1, df2, df3], ignore_index=True)

# Ergebnis kontrollieren
print("Gesamter Datensatz:", df.shape)
print("Spaltennamen:")
print(df.columns.tolist())
```

```python

# Fehlende Werte pro Spalte zählen
print("Fehlende Werte pro Spalte:")
print(df.isnull().sum())

# Anzahl der doppelten Zeilen
print("\nDoppelte Zeilen:", df.duplicated().sum())

# Datentypen anzeigen
print("\nDatentypen:")
print(df.dtypes)
```

```python

# Anzahl der Zeilen vor der Bereinigung
print("Vorher:", df.shape)

# Vollständig leere Zeilen entfernen
df = df.dropna(how="all").copy()

# Anzahl der Zeilen nach der Bereinigung
print("Nachher:", df.shape)

# Fehlende Werte erneut überprüfen
print("\nFehlende Werte:")
print(df.isnull().sum())

# Verbleibende Duplikate
print("\nDoppelte Zeilen:", df.duplicated().sum())
```

```python

# Fehlende Geschlechtsangaben ersetzen
df["gender"] = df["gender"].fillna("Unknown")

# Doppelte Zeilen entfernen
df = df.drop_duplicates().copy()

# Ergebnisse kontrollieren
print("Datensatz nach Bereinigung:", df.shape)

print("\nFehlende Werte:")
print(df.isnull().sum())

print("\nDoppelte Zeilen:", df.duplicated().sum())
```

```python

# Prozentzeichen entfernen
df["customer lifetime value"] = (
    df["customer lifetime value"]
    .astype(str)
    .str.replace("%", "", regex=False)
)

# In Zahlen umwandeln
df["customer lifetime value"] = pd.to_numeric(
    df["customer lifetime value"],
    errors="coerce"
)

# Fehlende Werte durch Median ersetzen
df["customer lifetime value"] = df["customer lifetime value"].fillna(
    df["customer lifetime value"].median()
)

# Ergebnis anzeigen
print(df["customer lifetime value"].dtype)
print(df["customer lifetime value"].isnull().sum())

df[["customer", "customer lifetime value"]].head()
```

```python

# Datensatz überprüfen
print("Anzahl Zeilen und Spalten:", df.shape)

print("\nFehlende Werte:")
print(df.isnull().sum())

print("\nDoppelte Zeilen:")
print(df.duplicated().sum())

print("\nDatentypen:")
print(df.dtypes)

# Bereinigten Datensatz speichern
df.to_csv("insurance_data_cleaned.csv", index=False)

print("\nDatensatz erfolgreich gespeichert!")
```

```python

# Beschwerdeanzahl in Zahlen umwandeln
df["number of open complaints"] = pd.to_numeric(
    df["number of open complaints"]
    .astype(str)
    .str.split("/")
    .str[0],
    errors="coerce"
)

# Ergebnis kontrollieren
print("Datentyp:", df["number of open complaints"].dtype)
print("Fehlende Werte:", df["number of open complaints"].isnull().sum())

# Bereinigte Datei erneut speichern
df.to_csv("insurance_data_cleaned.csv", index=False)

print("Datei aktualisiert!")
```

```python
df_marketing = pd.read_csv(
    "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis_clean.csv"
)

print(df_marketing.shape)
print(df_marketing.columns.tolist())

df_marketing.head()
```

```python

# Pivot-Tabelle nach Vertriebskanal erstellen
pivot_sales = df_marketing.pivot_table(
    index="sales_channel",
    values="monthly_premium_auto",
    aggfunc="sum"
)

# Auf zwei Nachkommastellen runden
pivot_sales = pivot_sales.round(2)

# Nach Umsatz absteigend sortieren
pivot_sales = pivot_sales.sort_values(
    by="monthly_premium_auto",
    ascending=False
)

# Ergebnis anzeigen
pivot_sales
```

### Analysis – Sales Channel Performance

The pivot table shows that the Agent sales channel generates the highest total monthly premium revenue (386,335), followed by Branch (280,953), Call Center (197,970), and Web (151,511).

The Agent channel contributes the most to the company's monthly premium revenue, while Web generates the lowest amount.

This suggests that agent-based sales are particularly important for the company. However, further analysis of customer numbers and average premiums would be necessary to evaluate the efficiency of each channel.

```python

# Durchschnittlichen Customer Lifetime Value berechnen
pivot_clv = df_marketing.pivot_table(
    index="gender",
    columns="education",
    values="customer_lifetime_value",
    aggfunc="mean"
)

# Auf zwei Nachkommastellen runden
pivot_clv = pivot_clv.round(2)

# Ergebnis anzeigen
pivot_clv
```

### Analysis – Customer Lifetime Value by Gender and Education

The pivot table shows differences in average customer lifetime value (CLV) across gender and education levels.

Female customers with a High School or Below education have the highest average CLV (8,675.22), while female customers with a Doctor degree have the lowest (7,328.51).

Male customers with a Master's degree have an average CLV of 8,168.83, slightly higher than female customers with the same education level (8,157.05).

Overall, there is no consistent relationship between higher education levels and higher customer lifetime value. Gender differences also vary depending on education level.

Further analysis would be needed to understand the reasons behind these differences.

```python

# Pivot-Tabelle mit Beschwerden erstellen
pivot_complaints = df_marketing.pivot_table(
    index="policy_type",
    columns="month",
    values="number_of_open_complaints",
    aggfunc="sum",
    fill_value=0
)

# Tabelle ins Long Format umwandeln
complaints_long = pivot_complaints.stack().reset_index()

# Spalten umbenennen
complaints_long.columns = [
    "Policy Type",
    "Month",
    "Total Complaints"
]

# Nach Anzahl der Beschwerden sortieren
complaints_long = complaints_long.sort_values(
    by="Total Complaints",
    ascending=False
)

# Ergebnis anzeigen
complaints_long.head(15)
```

```python

# Datentyp und Werte der Beschwerden überprüfen
print("Datentyp:")
print(df_marketing["number_of_open_complaints"].dtype)

print("\nBeispielwerte:")
print(df_marketing["number_of_open_complaints"].head(10))

print("\nStatistische Übersicht:")
print(df_marketing["number_of_open_complaints"].describe())
```

```python

# Beschwerdedaten überprüfen
complaints = df_marketing["number_of_open_complaints"]

# Häufigkeit der verschiedenen Werte anzeigen
print(complaints.value_counts().sort_index())

# Anzahl der nicht ganzzahligen Werte
invalid = complaints.notna() & (complaints % 1 != 0)

print("\nNicht ganzzahlige Werte:", invalid.sum())
```

```python

# Kopie des Datensatzes erstellen
df_bonus = df_marketing.copy()

# Nicht ganzzahlige Beschwerdewerte als fehlend markieren
complaints = df_bonus["number_of_open_complaints"]

df_bonus.loc[
    complaints.notna() & (complaints % 1 != 0),
    "number_of_open_complaints"
] = float("nan")

# Pivot-Tabelle erstellen
pivot_complaints = df_bonus.pivot_table(
    index="policy_type",
    columns="month",
    values="number_of_open_complaints",
    aggfunc="sum"
)

# Ins Long Format umwandeln
complaints_long = pivot_complaints.stack().reset_index()

# Spalten umbenennen
complaints_long.columns = [
    "Policy Type",
    "Month",
    "Total Complaints"
]

# Nach Beschwerden sortieren
complaints_long = complaints_long.sort_values(
    by="Total Complaints",
    ascending=False
)

# Ergebnis anzeigen
complaints_long
```

### Bonus Challenge – Complaints Analysis

The analysis shows that Personal Auto has the highest total number of open complaints in both January (1,635) and February (1,363).

Corporate Auto has considerably fewer complaints, with 415 in January and 361 in February. Special Auto has the lowest number of complaints, with 84 in January and 91 in February.

Overall, the total number of complaints decreased from January to February.

For this analysis, 633 non-integer values were excluded because they appear to represent imputed missing values rather than actual complaint counts.

These results show the total number of reported open complaints, but do not account for differences in the number of customers or policies across insurance categories.
