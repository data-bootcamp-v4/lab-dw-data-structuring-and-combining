In this challenge, we will be working with the customer data from an insurance company, as we did in the two previous labs. The data can be found here:

Note that you'll need to clean and format the new data.

### Strategy for the solution:

**Apply the cleaning function to the new combined dataset:**
- load the three raw CSV files
- merge them,
- cleaning the large dataset

```python
# Your code goes here

import numpy as np
import pandas as pd

df1 = pd.read_csv(
    'https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv'
)
df2 = pd.read_csv(
    'https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file2.csv'
)
df3 = pd.read_csv(
    'https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file3.csv'
)
```

```python
print("Shape of dataframes:")
display(f"Shape df1: {df1.shape}")
display(f"Shape df2: {df2.shape}")
display(f"Shape df3: {df3.shape}")


print("\nInfo about dataframes:")

display(f"Info df1: {df1.info()}")
display(f"Info df2: {df2.info()}")
display(f"Info df3: {df3.info()}")
```

```python
# Dealing with the column names:

for df in [df1, df2, df3]:
  df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')


df1 = df1.rename(columns={'st': 'state'})
df2 = df2.rename(columns={'st': 'state'})

print("\nColumns of df1:", df1.columns.tolist())
print("\nColumns of df2:", df2.columns.tolist())
print("\nColumns of df3:", df3.columns.tolist())
```

```python
# Combine the 3 dataframe since the names of the columns are the same
combined_df = pd.concat([df1, df2, df3], ignore_index=True)

print("\nColumns of combined df:", combined_df.columns.tolist())
print("\n")
display(f"Shape of combined df: {combined_df.shape}")
print("\n")
display(f"Info of combined df: {combined_df.info()}")
```

```python
# customer_lifetime_value - delete % and convert to float
combined_df['customer_lifetime_value'] = (
    combined_df['customer_lifetime_value']
        .astype(str)
        .str.replace('%', '', regex=False)
)

combined_df['customer_lifetime_value'] = pd.to_numeric(
    combined_df['customer_lifetime_value'], errors='coerce'
)

display(combined_df[['customer_lifetime_value']].info())
```

```python
# number_of_open_complaints - cleaning and convert to integer
combined_df['number_of_open_complaints'] = (
    combined_df['number_of_open_complaints'].astype(str).str.split('/').str[-1]
)
combined_df['number_of_open_complaints'] = pd.to_numeric(
    combined_df['number_of_open_complaints'], errors='coerce'
)
combined_df['number_of_open_complaints'] = (
    combined_df['number_of_open_complaints'].fillna(0).astype(int)
)

display(combined_df[['number_of_open_complaints']].info())
```

```python
# 3. Standardization of categories (Gender, State, Education)
# print(combined_df['vehicle_class'].unique())


# [nan, 'F', 'M', 'Femal', 'Male', 'female']
# ['Washington',    'Arizona',     'Nevada', 'California',     'Oregon',       'Cali',         'AZ',         'WA',          nan]
# [              'Master',             'Bachelor', 'High School or Below','College','Bachelors','Doctor', nan]
# ['Four-Door Car',  'Two-Door Car',           'SUV',    'Luxury SUV','Sports Car',    'Luxury Car']


# Gender - and insert U, as unkonwn if we have missing data
# I'm not delete these lines because the insured event, the claim, still exists.

combined_df['gender'] = combined_df['gender'].replace({
     'Femal': 'F',
     'female': 'F',
     'Male': 'M',
})

combined_df['gender'] = combined_df['gender'].fillna('U')


combined_df['state'] = combined_df['state'].replace(
     {'AZ': 'Arizona', 'Cali': 'California', 'WA': 'Washington'}
)

combined_df['education'] = combined_df['education'].replace(
     {'Bachelors': 'Bachelor'}
 )

combined_df['vehicle_class'] = combined_df['vehicle_class'].replace(
     {
         'Luxury SUV': 'Luxury',
         'Sports Car':'Luxury',
         'Luxury Car':'Luxury'
     }
 )

print(combined_df['gender'].unique())
print(combined_df['state'].unique())
print(combined_df['education'].unique())
print(combined_df['vehicle_class'].unique())
```

```python
# 4. Clean duplicated and empty rows 
combined_df = combined_df.dropna(how='all')

combined_df = combined_df.drop_duplicates().reset_index(drop=True)

display(f"Shape of combined df: {combined_df.shape}")
```

```python
# 4. Round 2 the float64 types columns
combined_df['customer_lifetime_value','income','monthly_premium_auto','total_claim_amount'] = combined_df[
    'customer_lifetime_value'
].round(2)

display(combined_df)
```

### Dealing with Null values

```python
print("Missing values per column:")
print('\n')
print(combined_df.isnull().sum())
```

```python
# Since 'customer' appears to be a unique value, I discard the row if it is empty.

combined_df = combined_df.dropna(subset=['customer'])
print(combined_df.isnull().sum())
```

```python
# check the customer where the customer_lifetime value is Nan:

print(combined_df[combined_df['customer_lifetime_value'].isna()].head())
```

This shows that these are genuine customers who have an income, pay a monthly insurance premium (`monthly_premium_auto`), and have even filed claims (`total_claim_amount`).
This indicates that these are genuine customers who have an income, pay a monthly insurance premium (`monthly_premium_auto`), and have even filed insurance claims (`total_claim_amount`).
This means that the `NaN` value here does not represent 0, but rather clearly points to data loss or a measurement error (missing data). 
So I deleted these rows.

```python
combined_df = combined_df.dropna(subset=['customer_lifetime_value'])
print(combined_df.isnull().sum())
```

### Save the clean dataframes

```python
combined_df.to_csv('combined_df.csv', index=False)
print("\nThe cleaned and formatted dataset saved:'combined_df.csv'")
```

# Challenge 2: Structuring Data

In this challenge, we will continue to work with customer data from an insurance company, but we will use a dataset with more columns, called marketing_customer_analysis.csv, which can be found at the following link:

https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis_clean.csv

This dataset contains information such as customer demographics, policy details, vehicle information, and the customer's response to the last marketing campaign. Our goal is to explore and analyze this data by performing data cleaning, formatting, and structuring.

```python
import numpy as np
import pandas as pd

url ="https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis_clean.csv"
marketing = pd.read_csv(url)


marketing.shape
display(marketing)

marketing.info()
```

### Column Names Operations

```python
marketing.columns = (
    marketing.columns
    .str.strip()        
    .str.lower()        
    .str.replace(" ", "_")
    .str.replace(":", "")
    
)

marketing = marketing.rename(columns={
    "employmentstatus": "employment_status"
})


marketing = marketing.drop(columns=['unnamed_0'])
# The column named `unnamed:_0` is most likely an index left over from a previous save operation
# as it does not contain any new business-relevant information, I removed it

display(marketing.columns.tolist())
```

### Cleaning invalid values

```python
display(marketing['effective_to_date'].unique())
display(marketing['effective_to_date'].nunique())

# round 2 decimal:
marketing['customer_lifetime_value'] = marketing['customer_lifetime_value'].round(2)
marketing['total_claim_amount'] = marketing['total_claim_amount'].round(2)
```

### Dtype operations

```python
marketing['months_since_last_claim'] = marketing['months_since_last_claim'].astype(int)
marketing['number_of_open_complaints'] = marketing['number_of_open_complaints'].astype(int)
marketing['effective_to_date'] = pd.to_datetime(marketing['effective_to_date'])
# now the format is: 2011-01-30 00:00:00, but I do not want to make more modification, 
# since as I understand if I make more modification (slice, forexample) it can be cause problems
# and this format does not effect the analyzation of data 

print(marketing.dtypes)
```

### Null Values

```python
print("Missing values per column:")
print(marketing.isnull().sum())

# since everywhere is 0, no action needed
```

### Duplicates

```python
duplicates_count = marketing.duplicated().sum()
print(f"Number of duplicated rows before cleaning: {duplicates_count}")
```

```python
# Number of duplicated rows before cleaning: 460

marketing = marketing.drop_duplicates(keep='first').reset_index(drop=True)
duplicates_count = marketing.duplicated().sum()
print(f"Number of duplicated rows after cleaning: {duplicates_count}")
```

```python
marketing.to_csv('marketing_cleaned_data.csv', index=False)
print("\nThe cleaned and formatted dataset saved:'marketing_cleaned_data.csv'")
```

1. You work at the marketing department and you want to know which sales channel brought the most sales in terms of total revenue. Using pivot, create a summary table showing the total revenue for each sales channel (branch, call center, web, and mail).
Round the total revenue to 2 decimal points.  Analyze the resulting table to draw insights.

2. Create a pivot table that shows the average customer lifetime value per gender and education level. Analyze the resulting table to draw insights.

```python
# 1. Pivot table for aggregation by sales_channel

revenue_by_channel = (
    marketing.pivot_table(
        index='sales_channel', values='customer_lifetime_value', aggfunc='sum'
    )
    .sort_values(by='customer_lifetime_value', ascending=False)
    .round(2)
)


revenue_by_channel.columns = ['total_revenue']

print('Revenue by sales channels:\n')

display(revenue_by_channel)
```

### Revenue by sales channels

* Dominance of Agent Channel: The Agent channel generates the highest revenue amount by far (over 31,7 million), indicating that personalized, human-driven sales and customer management handle the largest volume of high-value policies or claims.

* Traditional vs. Digital Channels: Traditional offline channels (Agent and Branch) account for the vast majority of the total claim amounts, significantly outperforming remote and digital touchpoints.

* Lower Digital Engagement: The Web channel records the lowest total claim amount (~12 million), suggesting lower customer adoption, smaller policy sizes, or simpler products being processed online compared to traditional methods.

* Strategic Recommendation: Management should investigate ways to boost digital channel adoption (Web) to capture more online market share, while ensuring continued support for the high-performing Agent and Branch networks.

```python
# 2. Pivot Table: Average Customer Lifetime Value by Gender and Education Level

clv_by_gender_edu = (
    marketing.pivot_table(
        index='education',
        columns='gender',
        values='customer_lifetime_value',
        aggfunc='mean',
    )
    .round(2)
)

print('Average Customer Lifetime Value by Gender and Education Level:\n')
display(clv_by_gender_edu)
```

### Average Customer Lifetime Value (CLV) by Gender and Education Level

Most Valuable Segment (Secondary Education):
- Women with a "High School or Below" education level have the highest average customer lifetime value (8,617.79 USD).
- Men in this category also show a high average (8,159.38 USD); consequently, this education level represents the most valuable customer segment for the insurance company overall.

Higher Education (Master’s & College):
- For individuals with a Master’s (MSc) degree, the CLV is balanced and above average for both genders (~8,200 USD).
- In the "College" category, men exhibit a notably higher value (8,075.26 USD) compared to women (7,757.32 USD).

Lowest Customer Lifetime Value (Doctoral Degree):
- Customers with a doctoral (PhD) degree have the lowest average CLV across both genders (~7,370 – 7,415 USD).

Gender differences:
- The disparity between genders is minimal (with a difference of 1–4% in most categories). Educational attainment is a significantly stronger indicator than gender when estimating customer value.


### Insight: 
The pivot table reveals that it is not those with the highest level of education (Doctorate) who generate the highest Customer Lifetime Value, but rather customers with a secondary education (High School or Below). Since the difference between genders is negligible, it is more effective to focus on educational attainment and income segments—rather than gender—when targeting marketing campaigns.
