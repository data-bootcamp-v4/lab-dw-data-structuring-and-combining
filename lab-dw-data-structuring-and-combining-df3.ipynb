```python
import pandas as pd 
import numpy as np
_pd3 = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file3.csv")
df3 = _pd3.copy()
df3.head(5)
```

```python
df3.columns
```

```python
df3.info()
```

```python
# Formating the column names
df3 = df3.rename(columns={col:col.replace(' ','_').lower() for col in df3.columns})
df3
```

```python
# display the indexes of the columns to change the order
for i, col in enumerate(df3.columns): print(i, col)
```

```python
df3.columns
```

```python
# order the columns in the same order than df1
new_order = [ 
    'customer', 
    'state', 
    'gender', 
    'education', 
    'customer_lifetime_value', 
    'income', 
    'monthly_premium_auto', 
    'number_of_open_complaints', 
    'policy_type', 
    'vehicle_class', 
    'total_claim_amount' 
] 
df3 = df3[new_order]
```

```python
# display the indexes of the columns to see the new order
for i, col in enumerate(df3.columns): print(i, col)
```

```python
df3.gender.unique()
```

```python
list(df3.state.unique())
```

```python
df3.education.unique()
```

```python
# In vehicle class, "Sports Car", "Luxury SUV" and "Luxury Car" replaced by "Luxury"
list(df3.vehicle_class.unique())
```

```python
vehicle_class_map ={
    'Luxury SUV':'Luxury',
    'Sports Car':'Luxury',
    'Luxury Car':'Luxury',
}
df3.vehicle_class = df3.vehicle_class.replace(vehicle_class_map)
```

```python
df3.customer_lifetime_value.unique()
```

```python
# convert Number of open complaints
df3.number_of_open_complaints.unique()
```

#### Dealing with Null values

```python
# Display the columns with the number of nan
df3.isna().sum()
```

```python
df3.describe()
```

```python
df3.dtypes
```

```python
# convert all the numeric variables to integers
numerical_values = [
    'customer_lifetime_value',
    'income', 
    'monthly_premium_auto', 
    'total_claim_amount'
]

df3[numerical_values]= df3[numerical_values].astype(int)
```

```python
df3.dtypes
```

#### Dealing with duplicated

```python
df3.duplicated().sum()
```

```python
# Dropping all duplicate rows
df3.drop_duplicates(inplace=True)
```

```python
df3.shape
```

```python
# reset the index to ensure consistency, after dropping duplicates
df3_without_duplicates = df3.copy()
df3_without_duplicates = df3_without_duplicates.drop_duplicates().reset_index(drop=True)
```

```python
# Save the cleaned dataset to a new CSV file
df3.to_csv("insurance_clean_file3.csv")
```
